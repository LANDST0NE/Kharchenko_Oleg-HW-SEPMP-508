```python
import networkx as nx
import matplotlib.pyplot as plt

G = nx.Graph()
edges = [
    ('A', 'B', 1),
    ('A', 'C', 1),
    ('A', 'D', 2),
    ('A', 'E', 3),
    ('C', 'D', 2),
    ('D', 'E', 4),
    ('B', 'E', 5),
]

G.add_weighted_edges_from(edges)

plt.figure(figsize=(6, 5))
pos = nx.spring_layout(G)

nx.draw(G, pos, with_labels=True,
        node_color='lightblue',
        edge_color='gray',
        width=2,
        node_size=800)
edge_labels = nx.get_edge_attributes(G, 'weight')
nx.draw_networkx_edge_labels(G, pos, edge_labels=edge_labels, font_size=12)

plt.title("Домашнее задание 1", fontsize=16)
plt.show()
```

## Домашнее задание 2




```python
import networkx as nx
import matplotlib.pyplot as plt

G = nx.DiGraph()
edges = [
    ('А', 'Б'),
    ('Б', 'Е'),
    ('Е', 'И'),
    ('А', 'В'),
    ('Б', 'В'),
    ('В', 'Е'),
    ('Е', 'Ж'),
    ('В', 'Ж'),
    ('Ж', 'И'),
    ('А', 'Г'),
    ('Г', 'В'),
    ('В', 'Ж'),
    ('Г', 'Ж'),
    ('А', 'Д'),
    ('Д', 'Г'),
    ('Д', 'Ж'),
    ('Д', 'З'),
    ('З', 'Ж'),
    ('З', 'И')

]

G.add_edges_from(edges)

plt.figure(figsize=(6, 5))
pos = nx.spring_layout(G)

nx.draw(G, pos, with_labels=True,
        node_color='lightblue',
        edge_color='gray',
        width=2,
        node_size=800)
edge_labels = nx.get_edge_attributes(G, 'weight')
nx.draw_networkx_edge_labels(G, pos, edge_labels=edge_labels, font_size=12)

plt.title("Домашнее задание 2", fontsize=16)
plt.show()
```

```python
def dfs(graph, start, end, path=None, all_paths=None):
    if path is None:
        path = []
    if all_paths is None:
        all_paths = []

    path = path + [start]  # Добавляем текущую вершину в путь

    if start == end:
        all_paths.append(path)  # Путь найден
    else:
        for neighbor in graph.successors(start):
            if neighbor not in path:  # Проверяем только текущий путь!
                dfs(graph, neighbor, end, path, all_paths)

    return all_paths

paths_from_A_to_V = dfs(G, 'А', 'В')
paths_from_V_to_I = dfs(G, 'В', 'И')

total = len(paths_A_V) * len(paths_V_I)

print(f"Пути А → В: {len(paths_from_A_to_V)}")
print(f"Пути В → И: {len(paths_from_V_to_I)}")
print(f"В итоге количество различных путей из города А в город И, проходящих через город В, то есть ответ на задачу: {total}")
```
